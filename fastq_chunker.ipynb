```python
import os
import glob
import subprocess
import numpy as np
import random
```

```python
# dpath = "/nfs/turbo/umms-indikar/shared/projects/hybrid_reprogramming/data/fastq/biolegend_library/"
dpath = "/nfs/turbo/umms-indikar/shared/projects/cell_cycle/data/sc_cell_cycle_2024/barcodes/fastq_pass/"

# file_list = glob.glob(f"{dpath}/*L2*/*.fastq.gz")
file_list = glob.glob(f"{dpath}/*.fastq.gz")
print(f"{len(file_list)=}")
random.shuffle(file_list)

file_list[:10]
```

```python
# output_path = "/nfs/turbo/umms-indikar/shared/projects/hybrid_reprogramming/data/fastq/biolegend_library/merged_TSB_L2/"
output_path = "/nfs/turbo/umms-indikar/shared/projects/cell_cycle/data/sc_cell_cycle_2024/barcodes/merged_chunked_tsb/"

chunks = 25
chunked = np.array_split(file_list, chunks)
print(f"{len(chunked)=}")


for i, flist in enumerate(chunked):
    print(f"Working chunk {i}...")
    outpath = f"{output_path}chunk_{i+1}.fastq.gz"
    flist_str = " ".join(flist)
    command = " ".join(['cat', flist_str, '>', outpath])
    subprocess.run(command, shell=True)

print('done')
```
