```python
import pandas as pd
```

```python
fpath = "/scratch/indikar_root/indikar1/shared_data/library_2_tsb/feature_barcodes/fbc_map.csv"

df = pd.read_csv(fpath)
print(f"{df.shape=}")
df['pred_split'] = df['pred'].apply(lambda x: "-".join(x.split("_")[2:]))
df.head()
```

```python
df['pred_split'].value_counts()
```

```python
df['pred_split'].value_counts(normalize=True)
```

```python
gx = df.groupby('cell_barcode')['pred_split'].value_counts().reset_index()
gx = pd.pivot_table(gx, index='cell_barcode', columns='pred_split', values='count', aggfunc='sum')
print(f"{gx.shape=}")
gx['pred'] = gx.idxmax(axis=1)
gx.head()
```

```python
(gx[['MYOD', 'PRRX1', 'PRRX1-MYOD']].std(axis=1) < 10).sum()
```

```python
gx['pred'].value_counts()
```
