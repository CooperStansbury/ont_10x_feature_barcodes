```python
import pandas as pd
import numpy as np
```

```python
fpath = "/scratch/indikar_root/indikar1/shared_data/bone_marrow_tsb/feature_barcodes/fbc_map.csv"
df = pd.read_csv(fpath)

print(df.columns)

# Step 1: Identify HTO columns
hto1_cols = [col for col in df.columns if col.endswith('HTO_1')]
hto2_cols = [col for col in df.columns if col.endswith('HTO_2')]

# Step 2: Compute raw sums per row
df = pd.DataFrame({
    'cell_barcode': df['cell_barcode'],
    'HTO_1': df[hto1_cols].sum(axis=1, skipna=True),
    'HTO_2': df[hto2_cols].sum(axis=1, skipna=True),
})

# Step 3: Group by barcode and sum
df = df.groupby('cell_barcode', as_index=False).sum()

df['pred'] = df[['HTO_1', 'HTO_2']].idxmax(axis=1)

# Result
print(df.head().to_string())
```

```python
counts = df['pred'].value_counts()
percentages = df['pred'].value_counts(normalize=True) * 100

summary = pd.DataFrame({
    'Count': counts,
    'Percent': percentages.round(2)
})

print("HTO assignment prediction counts and percentages:")
print(summary.to_string())
```
