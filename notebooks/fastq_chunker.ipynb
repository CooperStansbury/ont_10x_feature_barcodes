```python
import os
import glob
import subprocess
import numpy as np
import random
```

```python
# dpath = "/nfs/turbo/umms-indikar/shared/projects/hybrid_reprogramming/data/fastq/biolegend_library/"
# dpath = "/nfs/turbo/umms-indikar/shared/projects/cell_cycle/data/sc_cell_cycle_2024/barcodes/fastq_pass/"
dpath = "/nfs/turbo/umms-indikar/shared/projects/bone_marrow/data/feature_barcodes/F2HS21BM_20250509/"

# file_list = glob.glob(f"{dpath}/*L2*/*.fastq.gz")
file_list = glob.glob(f"{dpath}/*.fastq.gz")
print(f"{len(file_list)=}")
random.shuffle(file_list)

file_list[:10]
```

```python
%%time
# Output path
output_path = "/nfs/turbo/umms-indikar/shared/projects/bone_marrow/data/feature_barcodes/merged_tsb/"

# Split into chunks
chunks = 25
chunked = np.array_split(file_list, chunks)
print(f"Total files: {len(file_list)}")
print(f"Splitting into {chunks} chunks...")

# Process each chunk
for i, flist in enumerate(chunked, 1):
    print(f"\n[Chunk {i}/{chunks}]")
    print(f"  → Number of files in this chunk: {len(flist)}")
    print(f"  → Saving to: chunk_{i}.fastq.gz")
    
    outpath = f"{output_path}chunk_{i}.fastq.gz"
    
    flist_str = " ".join(flist)
    command = f"cat {flist_str} > {outpath}"
    subprocess.run(command, shell=True)

print("\nAll chunks processed. Done.")
```

```python
file_list = glob.glob(f"{output_path}/*.fastq.gz")

for f in file_list:
    basename = os.path.basename(f).replace(".fastq.gz", "")
    print(f"{basename},{f}")
```
